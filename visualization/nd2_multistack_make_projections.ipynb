# Quick projections of multichannel 3D images (multistack ```.nd2``` Version) 

## Input

This recipe expects an input folder containing multistack, 3D & multichannel ```.nd2``` files of shape (stack, channel, z, y, x).

**NOTE:** single channel images are untested at the moment.

## Output

For each channel of each ```.nd2``` image in the input folder, a ```.png``` image showning z-,y- and x-projections will be created in the output folder. The name of output files will be ```{input file name (without ending)}_stack{stack index}_{channel name}_projected.png```

````python
import warnings
import itertools
from pathlib import Path
from skimage.io import imsave
from projection import load_multistack_multichannel_nd2, get_orthogonal_projections_8bit

# path containing files to visualize
in_path = Path('/data/agl_data/NanoFISH/Gabi/GS076_Fbn-L_Fbn-R/20230920_run1_sd/raw/')

# default: put results in subdirectory called 'projections'
out_path = in_path / 'projections'

# Intensity range for levelling
# Default: auto -> level based on quantiles 
intensity_range = 'auto'
auto_range_quantiles = (0.02, 0.9995)

# Alternatively, specify range directly
# intensity_range = (500, 1500)

# Type of projection, can be min/max/mean
projection_type='max'
````

````python
# get all nd2 files in in_path
in_files = sorted(list(Path(in_path).glob('*.nd2')))

in_files
````

````python
# make folder for results if it does not exist
if not out_path.exists():
    out_path.mkdir(parents=True)

for in_file in in_files:
    
    print(in_file)

    # read images, channel names, pixel sizes
    try:
        imgs, channel_names, pixel_size = load_multistack_multichannel_nd2(in_file)
    except:
        print(f'error loading file {in_file}, skipping')
        continue

    channel_names_i = itertools.cycle(channel_names)
    
    # do projections and save
    for idx, fov in enumerate(imgs):
    
        for img, channel_name in zip(fov, channel_names_i):

            projections_8bit = get_orthogonal_projections_8bit(img, pixel_size,
                    projection_type=projection_type, intensity_range=intensity_range, auto_range_quantiles=auto_range_quantiles)

            # make filepath for output
            outfile = out_path / (in_file.stem + f'_stack{idx}_{channel_name}_projected.png')

            # catch low contrast warning
            with warnings.catch_warnings():
                warnings.simplefilter('ignore', UserWarning)

                imsave(str(outfile), projections_8bit)

        print(f'saved projections of {str(in_file)}.')
````
