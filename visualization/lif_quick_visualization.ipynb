# Quick visualization of LIF files

## Input

This recipe expects a base input folder (plus optional raw data subfolder) containing ```.lif``` files from Leica microscopes. Each ```.lif``` file can contain multiple images (e.g. single acquisitions and Z-stacks); each of them will be visualized separately.

## Output

For each image in each ```.lif``` file, this will generate 8bit visualizations and save them as PNG for quick visualization using e.g. standard file explorers. If an image is 3D, orthogonal projections (corrected for anisotropic voxel size) will be created. Multiple output files are created for additional dimensions in the image (e.g. multi-position, time-point, ...).

By setting ```do_color``` to ```True```, a single RGB composite (channel colors as in Fiji by default) is created per stack; with ```False```, one visualization per channel is written. By setting ```use_channel_names``` to ```True```, channels are labeled with their names (e.g. ```_chDAPI```) instead of integer indices (```_ch0```) in output filenames.

````python
from pathlib import Path

from calmutils.color.color import DEFAULT_COLOR_NAMES
from quick_visualization import quick_visualize_lif
````

````python
# specify the upper level folder
in_path = '/home/david/Desktop/symbiosis_analysis_guse'

# subfolder (e.g. "raw") containing lif files, leave blank if the files are in in_path
raw_subdir = ''

# folder to save results to (relative to in_path)
out_subdir = 'quick_visualization'

# Intensity range for leveling
# Default: 'auto' -> level based on quantiles
intensity_range = 'auto'
auto_range_quantiles = (0.02, 0.9995)

# Alternatively, specify range directly
# intensity_range = (500, 1500)

# Type of projection for 3D images, can be min/max/mean
projection_type = 'max'

# whether to produce color composites (True) or separate visualizations per channel (False)
do_color = False

# colors for each channel, the default setting is same as in Fiji
visualization_colors = DEFAULT_COLOR_NAMES
# or pick custom
# visualization_colors = ('green', 'red', 'blue')

# whether to label channels with their names (True) instead of integer indices (False) in filenames
use_channel_names = False

# dimension(s) holding the color channels for RGB composites
# (first one present in the image wins; 'C' for most images, 'S' for RGB-sampled images)
color_dimension_label = ('C', 'S')
````

````python
# get all lif files in input location, show for verification
in_files = sorted((Path(in_path) / raw_subdir).glob('[!.]*.lif'))
in_files
````

````python
quick_visualize_lif(
    in_path,
    raw_subdir=raw_subdir,
    out_subdir=out_subdir,
    intensity_range=intensity_range,
    auto_range_quantiles=auto_range_quantiles,
    projection_type=projection_type,
    do_color=do_color,
    color_dimension_label=color_dimension_label,
    visualization_colors=visualization_colors,
    use_channel_names=use_channel_names,
)
````
