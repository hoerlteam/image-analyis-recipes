# Resave images as TIFF (```.lif``` Version) 

## Input

This recipe expects a base input folder (plus optional raw data subfolder) containing ```.lif``` files from Leica microscopes. Each ```.lif``` file can contain multiple images (e.g. single acquisitions and Z-stacks).

Optionally, you can specify along which dimensions to split output files (by default, it will be channels ```"C"```, time points ```"T"``` and positions ```"P"```) and a name for the output folder.

By setting ```z_projection_type``` to ```"max", "min", "sum"``` or ```"mean"```, this will save Z-projections of your images using the corresponding projection method. Leaving this parameter as ```None``` will skip projection.

## Output

For each coordinate of the split dimensions (channels, time points, ...) of each image in each ```.lif``` file in the input folder, a ```.tif``` file will be created in a output folder (which is placed in the base input directory). The name of output files will be ```{input file name (without ending)}_{image name}_ch{channel number}_tp{time point number}_pos{position number}.tif``` (only dimensions that are present in the image will show up in the output filename). Pixel sizes are stored in the TIFF metadata in microns.

````python
import os

from lif_resaving import resave_lif_flexible
````

````python
# specify the upper level folder
in_path = '/home/david/Desktop/symbiosis_analysis_guse'

# subfolder (e.g. "raw") containing lif files, leave blank if the files are in in_path
image_subfolder = ''

# folder to save results to
out_subfolder = 'tif'

# dimensions along which to split
split_dimensions=("C", "T", "P")

# projection (set to None to skip)
z_projection_type = None
# z_projection_type = "max"
````

````python
resave_lif_flexible(
    os.path.join(in_path, image_subfolder),
    out_path=os.path.join(in_path, out_subfolder),
    split_dimensions=split_dimensions,
    projection=z_projection_type
)
````
