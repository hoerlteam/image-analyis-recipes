# Resave images as TIFF (```.nd2``` Version) 

## Input

This recipe expects a base input folder (plus optional raw data subfolder) containing multidimensional ```.nd2``` files.

Optionally, you can specify along which dimensions to split input files (by default, it will be channels ```"C"```, time points ```"T"``` and positions ```"P"```) and a name for the output folder.

You can also restrict which raw files are resaved via ```file_pattern``` (a glob pattern or a list of patterns) and ```exclude_pattern```. This allows running two rounds into the same output folder, e.g. one round with ```file_pattern='*1_5*.nd2'``` for files taken with a manually flipped-in zoom lens. For those files set ```pixel_size_scale=[1, 1/1.5, 1/1.5]``` (lateral only - the z-spacing is unaffected by the zoom) so the true pixel size is written to the tiff metadata (the zoom is not reflected in the nd2 metadata); the second round then uses ```exclude_pattern='*1_5*.nd2'``` so it does not overwrite the zoom-corrected files.

By setting ```z_projection_type``` to ```"max", "min", "sum"``` or ```"mean"```, this will save Z-projections of your images using the corresponding projection method. Leaving this parameter as ```None``` will skip projection.

## Output

For each coordinate of the split dimensions (channels, time points, ...) of each ```.nd2``` image in the input folder, a ```.tif``` file will be created in a output folder (which is placed in the base input directory). The name of output files will be ```{input file name (without ending)}_ch{channel number}_tp{time point number}_pos{position number}.tif``` (only dimensions that are present in the file will show up in the output filename).

````python
import os

from nd2_resaving import resave_nd2_flexible
````

````python
# specify the upper level folder
in_path = '/Users/david/Desktop/scratch_data/20230616_SpinningDiskPSF'

# subfolder (e.g. "raw") containing nd2 files, leave blank if the files are in in_path
image_subfolder = ''

# folder to save results to
out_subfolder = 'tif'

# dimensions along which to split
split_dimensions=("C", "T", "P")

# projection (set to None to skip)
z_projection_type = None
# z_projection_type = "max"

# glob pattern(s) selecting the raw files to resave (list of patterns allowed)
file_pattern = '[!.]*.nd2'
# glob pattern(s) to exclude from file_pattern (e.g. handled in another round)
exclude_pattern = None

# multiply the metadata pixel size [z, y, x] by this - scalar or [z, y, x] sequence.
# Use [1, 1/1.5, 1/1.5] for files taken with a manually flipped-in 1.5x zoom lens
# (lateral only, z-spacing unchanged; not reflected in the nd2 metadata)
pixel_size_scale = 1.0
````

````python
resave_nd2_flexible(
    os.path.join(in_path, image_subfolder),
    file_pattern=file_pattern,
    exclude_pattern=exclude_pattern,
    out_path=os.path.join(in_path, out_subfolder),
    split_dimensions=split_dimensions,
    projection=z_projection_type,
    pixel_size_scale=pixel_size_scale
)
````
